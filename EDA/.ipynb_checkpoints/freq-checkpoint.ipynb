```python
query = """
SELECT
    time,
    current,
    freq,
    amp,
    dev
FROM fan_telemetry
WHERE time >= (
    SELECT MIN(time) + INTERVAL '10' HOUR
    FROM fan_telemetry
)
ORDER BY time ASC
"""

table = client.query(
    query=query,
    language="sql"
)

df = table.to_pandas()
df_clean = df[df["amp"] > 0].copy()

print("Data loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
df_clean = df[
    (df["amp"] > 0) &
    (df["freq"].notna()) &
    (df["amp"].notna())
].copy()

print("Raw shape:", df.shape)
print("Clean shape:", df_clean.shape)
print("Time range:", df_clean["time"].min(), "->", df_clean["time"].max())
```

```python
plt.figure(figsize=(14, 5))

sns.countplot(
    data=df_clean,
    x="freq",
    order=df_clean["freq"].value_counts().index[:15]
)

plt.title("Most Frequent Dominant Frequencies")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Number of observations")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FREQUENCY STABILITY OVER TIME
# ============================================================

df_freq = df_clean.copy()
df_freq = df_freq.set_index("time").sort_index()

# Statistics per 5-minute window
freq_window = (
    df_freq["freq"]
    .resample("5min")
    .agg(
        count="count",
        median="median",
        mean="mean",
        std="std",
        p05=lambda x: x.quantile(0.05),
        p95=lambda x: x.quantile(0.95)
    )
)

# Exclude windows with insufficient observations
freq_window = freq_window[
    freq_window["count"] >= 100
]

# Plot
fig, ax = plt.subplots(figsize=(16, 6))

# P05-P95 range
ax.fill_between(
    freq_window.index,
    freq_window["p05"],
    freq_window["p95"],
    alpha=0.2,
    label="P05-P95 range"
)

# Median frequency
ax.plot(
    freq_window.index,
    freq_window["median"],
    marker="o",
    markersize=3,
    linewidth=1.5,
    label="Median frequency"
)

ax.set_xlabel("Time")
ax.set_ylabel("Dominant Frequency (Hz)")
ax.set_title("Frequency Stability Over Time (5-minute windows)")

ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```
